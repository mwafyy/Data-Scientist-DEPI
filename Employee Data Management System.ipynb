```python
import csv
```

```python
class EmployeeManager:

    def __init__(self):
        self.employees = {}
        self.load_data()

    # read employees from csv when the program starts
    def load_data(self):
        try:
            with open("employees.csv", "r", newline="") as file:
                reader = csv.DictReader(file)

                for row in reader:
                    self.employees[row["ID"]] = {
                        "Name": row["Name"],
                        "Position": row["Position"],
                        "Salary": row["Salary"],
                        "Email": row["Email"]
                    }

        except FileNotFoundError:
            pass

    # save the current employees to the csv file
    def save_data(self):
        with open("employees.csv", "w", newline="") as file:
            fields = ["ID", "Name", "Position", "Salary", "Email"]

            writer = csv.DictWriter(file, fieldnames=fields)
            writer.writeheader()

            for emp_id, employee in self.employees.items():
                writer.writerow({
                    "ID": emp_id,
                    "Name": employee["Name"],
                    "Position": employee["Position"],
                    "Salary": employee["Salary"],
                    "Email": employee["Email"]
                })

    # add a new employee
    def add_employee(self):
        emp_id = input("Enter ID: ")

        if emp_id in self.employees:
            print("Employee ID already exists.")
            return

        name = input("Enter Name: ")
        position = input("Enter Position: ")

        while True:
            salary = input("Enter Salary: ")

            try:
                float(salary)
                break
            except ValueError:
                print("Salary must be a number.")

        email = input("Enter Email: ")

        self.employees[emp_id] = {
            "Name": name,
            "Position": position,
            "Salary": salary,
            "Email": email
        }

        self.save_data()
        print("Employee added successfully.")

    # display all employees
    def view_all(self):
        if len(self.employees) == 0:
            print("No employees found.")
            return

        for emp_id, employee in self.employees.items():
            print("\nID:", emp_id)
            print("Name:", employee["Name"])
            print("Position:", employee["Position"])
            print("Salary:", employee["Salary"])
            print("Email:", employee["Email"])

    # update employee information
    def update_employee(self):
        emp_id = input("Enter Employee ID: ")

        if emp_id not in self.employees:
            print("Employee not found.")
            return

        employee = self.employees[emp_id]

        name = input("New Name (leave empty to keep old): ")
        position = input("New Position (leave empty to keep old): ")
        salary = input("New Salary (leave empty to keep old): ")
        email = input("New Email (leave empty to keep old): ")

        if name != "":
            employee["Name"] = name

        if position != "":
            employee["Position"] = position

        if salary != "":
            while True:
                try:
                    float(salary)
                    employee["Salary"] = salary
                    break
                except ValueError:
                    print("Salary must be a number.")
                    salary = input("Enter Salary again: ")

        if email != "":
            employee["Email"] = email

        self.save_data()
        print("Employee updated successfully.")

    # delete employee
    def delete_employee(self):
        emp_id = input("Enter Employee ID: ")

        if emp_id in self.employees:
            del self.employees[emp_id]
            self.save_data()
            print("Employee deleted successfully.")
        else:
            print("Employee not found.")

    # search for one employee
    def search_employee(self):
        emp_id = input("Enter Employee ID: ")

        if emp_id in self.employees:
            employee = self.employees[emp_id]

            print("\nID:", emp_id)
            print("Name:", employee["Name"])
            print("Position:", employee["Position"])
            print("Salary:", employee["Salary"])
            print("Email:", employee["Email"])
        else:
            print("Employee not found.")
```

```python
manager = EmployeeManager()

while True:

    print("\nEmployee Data Management System")
    print("1. Add Employee")
    print("2. View All Employees")
    print("3. Update Employee")
    print("4. Delete Employee")
    print("5. Search Employee")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        manager.add_employee()

    elif choice == "2":
        manager.view_all()

    elif choice == "3":
        manager.update_employee()

    elif choice == "4":
        manager.delete_employee()

    elif choice == "5":
        manager.search_employee()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice. Please choose from 1 to 6.")
```
